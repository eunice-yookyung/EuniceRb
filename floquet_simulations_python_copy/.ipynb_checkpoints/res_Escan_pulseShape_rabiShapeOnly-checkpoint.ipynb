```python
# =============================================================
# resonance_Escan_profile.py
#
# Simplified, single-scan resonance profiler for photon-assisted tunneling
# between a chosen initial and target Fock state under lattice-depth
# modulation. Stripped-down version of resonance_Escan_pulseshape.py:
#
#   - NO iterative workflow (no CW time scans, no T_rabi fits).
#   - `pulse_npi` is replaced by `pulse_duration_ms`: the TOTAL wall-clock
#     duration of the pulse, in ms, for BOTH pulse shapes. This makes
#     (pulse_shape, pulse_duration_ms) a clean two-variable comparison:
#     the square and gaussian pulses occupy the same time window.
#   - Single frequency scan around the sideband guess; fit; plot panel (c)
#     only (simulation points, lineshape fit, fitted-center xline,
#     expected-center xline).
#
# Pulse definitions:
#   'square'   : constant modulation amplitude for t_pulse = pulse_duration_ms.
#                Lineshape model (fixed, KNOWN duration t_pulse):
#                  gen(x) = sqrt(OmegaR^2 + (x - x0)^2)          [Hz]
#                  P(x)   = A * OmegaR^2/gen^2 * sin^2(pi * gen * t_pulse)
#                (equivalent to the old n*pi form with n*pi = 2*pi*OmegaR*t_pulse,
#                 but with t_pulse fixed and OmegaR a free fit parameter.)
#
#   'gaussian' : V(t) = V0 * (1 + A(t) * sum_k alpha_k cos(om_k t + phi)),
#                A(t) = exp(-(t - t_c)^2 / (2 sigma^2)),
#                with the envelope truncated to the window [0, t_pulse]:
#                  t_pulse = pulse_duration_ms,  t_c = t_pulse/2,
#                  sigma   = t_pulse / (2 * gauss_trunc_sigma).
#                NOTE: because the window is fixed, the envelope AREA is
#                  integral A(t) dt = sigma*sqrt(2*pi)*erf(gauss_trunc_sigma/sqrt(2))
#                which is SMALLER than t_pulse (~0.42*t_pulse at n_trunc=3):
#                at equal duration the gaussian delivers less pulse area than
#                the square, so expect proportionally slower transfer.
#                Lineshape model: Gaussian peak
#                  P(x) = A * exp(-(x - x0)^2 / (2 w^2)),
#                accurate near the center (the only region that matters for x0).
#
# N_PERIODS_PER_RABI and TIME_SAMPLES_PER_PERIOD are accepted for call-site
# compatibility with resonance_Escan_pulseshape.py but are UNUSED here (there
# are no CW time scans).
# =============================================================

import itertools
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.optimize import curve_fit
from scipy.special import erf
import csv

twopi = 2.0 * np.pi

# =============================================================
# DEPTH -> (J, U) CONVERTERS  (module-level: no closure dependence)
# =============================================================
J_OUTPUT_UNITS = "Hz"
E_R_HZ = 1.0

# Found by fitting lattice depth to J for the Wannier functions (agrees w/ data).
def V_to_J(V):
    V = np.asarray(V, dtype=float)
    a = 284.4; b = -0.2922; c = 30.69; d = -0.1304   # Er in, Hz out
    return a * np.exp(b * V) + c * np.exp(d * V)

def J_angular(V):
    Jval = V_to_J(V)
    if J_OUTPUT_UNITS == "Hz":        return twopi * Jval
    elif J_OUTPUT_UNITS == "angular": return Jval
    elif J_OUTPUT_UNITS == "Er":      return twopi * E_R_HZ * Jval
    else: raise ValueError("J_OUTPUT_UNITS must be 'Hz', 'angular', or 'Er'")

# On-site interaction vs lattice depth (Er in, Hz out).
def U_of_V(V):
    V = np.asarray(V, dtype=float)
    a = 183.4; b = -56.83
    return a * V**0.25 + b

def U_angular(V):
    return twopi * U_of_V(V)


# =============================================================
# MAIN FUNCTION
# =============================================================
def resonance_Escan(state_init,
                    state_target,
                    V0_ER,
                    Delta_hz,
                    sideband,
                    alpha,
                    mu_offset_hz,
                    # Floquet-averaging controls
                    n_floquet_avg=1,
                    phi_floquet_avg=np.pi/2.0,
                    # Pulse controls
                    pulse_duration_ms=10.0,
                    pulse_shape="square",
                    gauss_trunc_sigma=3.0,
                    # scan / fit controls
                    N_PERIODS_PER_RABI=100,        # unused (kept for compatibility)
                    TIME_SAMPLES_PER_PERIOD=20,    # unused (kept for compatibility)
                    PULSE_SPAN_HZ=100.0,
                    PULSE_N_POINTS=None,
                    PULSE_SAMPLES_PER_PERIOD=40,
                    write_csv=True,
                    csv_path="simulation_results_Escan_profile.csv",
                    fig_path="resonance_Escan_profile.png",
                    show=True):
    """
    Single-scan resonance profiler. Returns a dict of results.

    Parameters
    ----------
    state_init   : list[int]   initial Fock state (length L)
    state_target : list[int]   target Fock state (length L, same N)
    V0_ER        : float       static lattice depth (Er)
    Delta_hz     : float       tilt per site (Hz)
    sideband     : int (+1/0/-1)  +1 -> guess = Delta + U_mean_0
                                   0 -> guess = Delta
                                  -1 -> guess = Delta - U_mean_0
    alpha        : list[float] depth modulation amplitude(s); must match #tones (=1)
    mu_offset_hz : list[float] per-site energy offset (Hz), length L, or empty []
    n_floquet_avg   : int      sample points per Floquet period used to form the
                               slow envelope. 1 -> stroboscopic; >1 -> boxcar.
    phi_floquet_avg : float    initial phase (radians) of the cosine drive.
    pulse_duration_ms : float  TOTAL duration of the pulse (ms), both shapes.
    pulse_shape  : str         'square' or 'gaussian' (see module header).
    gauss_trunc_sigma : float  gaussian window half-width in units of sigma;
                               sigma = t_pulse / (2*gauss_trunc_sigma).
    PULSE_N_POINTS : int/None  points in the frequency scan. If None:
                               'gaussian' -> 41 (single smooth peak);
                               'square'   -> scaled to resolve the sinc fringes,
                               ~8 points per fringe of width 1/t_pulse across
                               the scan span (minimum 41, forced odd).
    """

    # ----------------------------------------------------------------
    # Static lattice depth + drive amplitudes
    # ----------------------------------------------------------------
    V0    = float(V0_ER)
    alpha = np.atleast_1d(np.asarray(alpha, dtype=float))
    Delta = twopi * Delta_hz

    sideband = int(np.sign(sideband))

    # Floquet-averaging controls
    n_floquet_avg   = max(1, int(n_floquet_avg))
    phi_floquet_avg = float(phi_floquet_avg)

    # Pulse controls
    pulse_duration_ms = float(pulse_duration_ms)
    if pulse_duration_ms <= 0:
        raise ValueError("pulse_duration_ms must be > 0.")
    t_pulse = 1e-3 * pulse_duration_ms   # seconds

    pulse_shape = str(pulse_shape).lower()
    if pulse_shape not in ("square", "gaussian"):
        raise ValueError("pulse_shape must be 'square' or 'gaussian'; got %r."
                         % pulse_shape)
    gauss_trunc_sigma = float(gauss_trunc_sigma)
    if pulse_shape == "gaussian" and gauss_trunc_sigma <= 0:
        raise ValueError("gauss_trunc_sigma must be > 0.")

    if pulse_shape == "gaussian":
        gauss_sigma = t_pulse / (2.0 * gauss_trunc_sigma)
        # Envelope area within the truncated window (for reference/printout):
        gauss_area = (gauss_sigma * np.sqrt(2.0 * np.pi)
                      * erf(gauss_trunc_sigma / np.sqrt(2.0)))
    else:
        gauss_sigma = None
        gauss_area = t_pulse

    if PULSE_N_POINTS is None:
        if pulse_shape == "square":
            # Fringe spacing ~ 1/t_pulse; aim for ~8 points per fringe.
            n = int(np.ceil(8.0 * (2.0 * PULSE_SPAN_HZ) * t_pulse))
            PULSE_N_POINTS = max(41, n | 1)   # at least 41, force odd
        else:
            PULSE_N_POINTS = 41
    pulse_label = r"%.2f\ \mathrm{ms\ %s}" % (pulse_duration_ms, pulse_shape)

    state_init = list(state_init)
    L = len(state_init)
    N = sum(state_init)

    # ----------------------------------------------------------------
    # Mean / static interaction values from the depth converter
    # ----------------------------------------------------------------
    def _depth_trace(om, n_sub=10000):
        T = 1.0 / (om.min() / twopi)
        t = np.linspace(0.0, T, n_sub, endpoint=False)
        return V0 * (1.0 + np.sum(alpha[:, None]
                                  * np.cos(np.outer(om, t) + phi_floquet_avg), axis=0))

    om_rep = twopi * np.array([1.0])
    Vt_rep = _depth_trace(om_rep)
    U_mean_0 = U_of_V(Vt_rep).mean()
    U_V0     = float(U_of_V(V0))

    # ----------------------------------------------------------------
    # Drive frequency guess from the sideband sign
    # ----------------------------------------------------------------
    f_drive_hz_init = Delta_hz + sideband * U_mean_0
    if sum(state_target) != N:
        raise ValueError("init and target must share particle number N. "
                         "(target=%s, N=%d)" % (state_target, N))

    # ----------------------------------------------------------------
    # Per-site chemical potential (angular). Empty -> zeros.
    # ----------------------------------------------------------------
    mu_in = np.asarray(mu_offset_hz, dtype=float).ravel()
    if mu_in.size == 0:
        mu_site_hz = np.zeros(L)
    elif mu_in.size == L:
        mu_site_hz = mu_in.copy()
    else:
        raise ValueError("mu_offset_hz must be empty or length L=%d (got %d)."
                         % (L, mu_in.size))
    mu_site = twopi * mu_site_hz  # angular

    # ----------------------------------------------------------------
    # Fixed-N basis
    # ----------------------------------------------------------------
    def generate_basis(L, N):
        return [tuple(occ) for occ in itertools.product(range(N + 1), repeat=L)
                if sum(occ) == N]

    basis = generate_basis(L, N)
    index = {s: i for i, s in enumerate(basis)}
    dim   = len(basis)

    if tuple(state_init) not in index or tuple(state_target) not in index:
        raise ValueError("init/target state not present in the N=%d basis." % N)

    print("L = %d, N = %d, dim = %d" % (L, N, dim))
    print("init = %s, target = %s" % (tuple(state_init), tuple(state_target)))
    print("sideband = %+d -> f_drive_hz_init = %.2f Hz" % (sideband, f_drive_hz_init))
    print("static depth V0 = %.4g Er -> J(V0) = %.4f Hz" % (V0, J_angular(V0)/twopi))
    print("U_mean_0 = %.4f Hz, U(V0) = %.4f Hz" % (U_mean_0, U_V0))
    print("n_floquet_avg = %d, phi_floquet_avg = %.4f rad" % (n_floquet_avg, phi_floquet_avg))
    print("pulse: shape = %s, total duration = %.4f ms  (%d scan pts)"
          % (pulse_shape, pulse_duration_ms, PULSE_N_POINTS))
    if pulse_shape == "gaussian":
        print("gaussian: sigma = %.4f ms, envelope area = %.4f ms "
              "(= %.3f of the square area at equal duration)"
              % (1000*gauss_sigma, 1000*gauss_area, gauss_area/t_pulse))

    # ----------------------------------------------------------------
    # Hamiltonian pieces.
    # ----------------------------------------------------------------
    def build_static_diag():
        H = np.zeros((dim, dim), dtype=complex)
        site = np.arange(L)
        for a, state in enumerate(basis):
            n = np.array(state)
            tilt = -Delta * np.sum(site * n)
            chem = np.sum(mu_site * n)
            H[a, a] = tilt + chem
        return H

    def build_int_unit():
        # Diagonal: sum_i 0.5 * n_i (n_i - 1). Multiplied by U(t) in the RHS.
        H = np.zeros((dim, dim), dtype=complex)
        for a, state in enumerate(basis):
            n = np.array(state)
            H[a, a] = 0.5 * np.sum(n * (n - 1))
        return H

    def build_hop_unit():
        H = np.zeros((dim, dim), dtype=complex)
        for a, state in enumerate(basis):
            n = np.array(state)
            for j in range(L - 1):
                if n[j] > 0:
                    new = n.copy(); amp = np.sqrt(n[j]) * np.sqrt(n[j+1] + 1)
                    new[j] -= 1; new[j+1] += 1
                    H[index[tuple(new)], a] += -amp
                if n[j+1] > 0:
                    new = n.copy(); amp = np.sqrt(n[j+1]) * np.sqrt(n[j] + 1)
                    new[j+1] -= 1; new[j] += 1
                    H[index[tuple(new)], a] += -amp
        return H

    H_static   = build_static_diag()
    H_int_unit = build_int_unit()
    H_hop_unit = build_hop_unit()

    # ----------------------------------------------------------------
    # State and projector
    # ----------------------------------------------------------------
    psi0 = np.zeros(dim, dtype=complex)
    psi0[index[tuple(state_init)]] = 1.0
    P_target = np.zeros((dim, dim), dtype=complex)
    P_target[index[tuple(state_target)], index[tuple(state_target)]] = 1.0

    # ----------------------------------------------------------------
    # Evolution helper.
    #   `envelope` (optional callable t -> [0,1]) multiplies the modulation
    #   amplitude only; the static depth V0 is untouched.
    # ----------------------------------------------------------------
    def V_of_t(t, om, envelope=None):
        env = 1.0 if envelope is None else envelope(t)
        return V0 * (1.0 + env * np.sum(alpha * np.cos(om * t + phi_floquet_avg)))

    def evolve(samples_per_period, om, t_end, envelope=None):
        T_ref_local = 1.0 / (om.min() / twopi)
        n_per = max(1, int(round(t_end / T_ref_local)))
        ntimes = n_per * samples_per_period + 1
        def rhs(t, psi):
            Vt = V_of_t(t, om, envelope)
            Jt = J_angular(Vt)
            Ut = U_angular(Vt)
            H = H_static + Ut * H_int_unit + Jt * H_hop_unit
            return -1j * H.dot(psi)
        times = np.linspace(0.0, t_end, ntimes)
        sol = solve_ivp(rhs, (0.0, t_end), psi0, t_eval=times,
                        rtol=1e-9, atol=1e-9, method="RK45")
        return times, sol.y.T

    def ket(s):
        return r"|%s\rangle" % ",".join(str(x) for x in s)

    # ----------------------------------------------------------------
    # Lineshape models.
    #   square   : fixed-duration generalized-Rabi lineshape (sinc-like fringes)
    #   gaussian : Gaussian peak (the smooth envelope kills the side lobes)
    # ----------------------------------------------------------------
    def lineshape_square(x, A, OmegaR, x0):
        gen = np.sqrt(OmegaR**2 + (x - x0)**2)
        lor = OmegaR**2 / (OmegaR**2 + (x - x0)**2)
        return A * lor * np.sin(np.pi * gen * t_pulse)**2

    def lineshape_gauss(x, A, w, x0):
        return A * np.exp(-0.5 * ((x - x0) / w)**2)

    pulse_lineshape = lineshape_square if pulse_shape == "square" else lineshape_gauss

    def pulsed_population(f_drive):
        om = twopi * np.array([f_drive])
        if pulse_shape == "gaussian":
            t_c = 0.5 * t_pulse
            env = lambda t: np.exp(-0.5 * ((t - t_c) / gauss_sigma)**2)
            tt, pp = evolve(PULSE_SAMPLES_PER_PERIOD, om=om,
                            t_end=t_pulse, envelope=env)
        else:
            tt, pp = evolve(PULSE_SAMPLES_PER_PERIOD, om=om, t_end=t_pulse)
        pj = np.array([np.vdot(p, P_target.dot(p)).real for p in pp])
        # Reduce the final-time population using the same Floquet-averaging control.
        if n_floquet_avg <= 1:
            return pj[-1]
        wloc = min(int(n_floquet_avg), len(pj))
        kloc = np.ones(wloc) / wloc
        return np.convolve(pj, kloc, mode="valid")[-1]

    def fit_profile(f_center_guess, label=""):
        scan_freqs = np.linspace(f_center_guess - PULSE_SPAN_HZ,
                                 f_center_guess + PULSE_SPAN_HZ, PULSE_N_POINTS)
        pulse_pop = np.array([pulsed_population(f) for f in scan_freqs])
        f_peak = scan_freqs[np.argmax(pulse_pop)]

        if pulse_shape == "square":
            # OmegaR guess: assume the pulse is near a pi pulse -> OmegaR*t = 1/2.
            width_guess = 0.5 / t_pulse
        else:
            # Data-driven width: second moment of the (baseline-subtracted) peak.
            p_norm = pulse_pop - pulse_pop.min()
            if p_norm.sum() > 0:
                mu_w = np.sum(scan_freqs * p_norm) / p_norm.sum()
                width_guess = np.sqrt(np.sum((scan_freqs - mu_w)**2 * p_norm)
                                      / p_norm.sum())
            else:
                width_guess = 0.5 / t_pulse
            width_guess = max(width_guess, 1e-3)

        p0_line = [pulse_pop.max(), width_guess, f_peak]
        try:
            popt_line, pcov_line = curve_fit(pulse_lineshape, scan_freqs, pulse_pop,
                                             p0=p0_line, maxfev=40000)
            popt_line[1] = abs(popt_line[1])
            x0 = popt_line[2]
            perr = np.sqrt(np.diag(pcov_line))
            fit_ok = True
            wname = "OmegaR" if pulse_shape == "square" else "sigma_f"
            print("[profile %s] center x0 = %.3f Hz (+/- %.3f), %s = %.3f Hz"
                  % (label, x0, perr[2], wname, popt_line[1]))
        except RuntimeError:
            popt_line = p0_line; x0 = f_peak; fit_ok = False
            print("[profile %s] lineshape fit did not converge; using peak = %.3f Hz"
                  % (label, f_peak))
        return scan_freqs, pulse_pop, popt_line, x0, fit_ok

    def J_mean_0(om, n_sub=10000):
        T = 1.0 / (om.min() / twopi)
        t = np.linspace(0.0, T, n_sub, endpoint=False)
        Vt = V0 * (1.0 + np.sum(alpha[:, None]
                                * np.cos(np.outer(om, t) + phi_floquet_avg), axis=0))
        return V_to_J(Vt).mean()

    omega_init = twopi * np.array([f_drive_hz_init])
    print("J_mean_0      = %.4f Hz" % J_mean_0(omega_init))
    print("J(V0)         = %.4f Hz" % V_to_J(V0))

    # ----------------------------------------------------------------
    # Frequency profile scan + fit
    # ----------------------------------------------------------------
    print("\n=== frequency profile scan (%.2f ms %s pulse) ==="
          % (pulse_duration_ms, pulse_shape))
    scan_freqs, pulse_pop, popt_line, f_center_rabi, ok_line = \
        fit_profile(f_drive_hz_init, label="profile")

    detuning = f_center_rabi - f_drive_hz_init
    print("\n=== SUMMARY ===")
    print("pulse shape / duration = %s / %.4f ms" % (pulse_shape, pulse_duration_ms))
    print("f_drive_hz_init        = %.3f Hz" % f_drive_hz_init)
    print("f_center_rabi          = %.3f Hz" % f_center_rabi)
    print("detuning (fit - init)  = %+.3f Hz" % detuning)

    # ----------------------------------------------------------------
    # APPEND RESULTS TO RUNNING CSV DATABASE
    # ----------------------------------------------------------------
    if write_csv:
        FIELDS = ["V0_ER", "alpha", "N", "L", "sideband",
                  "n_floquet_avg", "phi_floquet_avg",
                  "pulse_duration_ms", "pulse_shape", "gauss_trunc_sigma",
                  "J_mean_0", "J(V0)", "U_mean_0", "U(V0)",
                  "Delta_hz",
                  "f_resonance_hz", "fit_width_hz", "detuning",
                  "initial_state", "final_state", "mu_offset_hz"]

        row = {
            "V0_ER":              V0,
            "alpha":              ";".join("%.6g" % a for a in alpha),
            "N":                  N,
            "L":                  L,
            "sideband":           sideband,
            "n_floquet_avg":      n_floquet_avg,
            "phi_floquet_avg":    phi_floquet_avg,
            "pulse_duration_ms":  pulse_duration_ms,
            "pulse_shape":        pulse_shape,
            "gauss_trunc_sigma":  gauss_trunc_sigma if pulse_shape == "gaussian" else "",
            "J_mean_0":           J_mean_0(omega_init),
            "J(V0)":              float(V_to_J(V0)),
            "U_mean_0":           U_mean_0,
            "U(V0)":              U_V0,
            "Delta_hz":           Delta_hz,
            "f_resonance_hz":     f_center_rabi,
            "fit_width_hz":       popt_line[1] if ok_line else "",
            "detuning":           detuning,
            "initial_state":      "".join(str(x) for x in state_init),
            "final_state":        "".join(str(x) for x in state_target),
            "mu_offset_hz":       ";".join("%.6g" % m for m in mu_site_hz),
        }

        write_header = (not os.path.exists(csv_path)) or os.path.getsize(csv_path) == 0
        with open(csv_path, "a", newline="") as f:
            writer = csv.DictWriter(f, fieldnames=FIELDS)
            if write_header:
                writer.writeheader()
            writer.writerow(row)
        print("appended results -> %s" % os.path.abspath(csv_path))

    # ----------------------------------------------------------------
    # PLOT: single panel (former panel (c))
    # ----------------------------------------------------------------
    amp_str = ", ".join("%.2f" % a for a in alpha)
    Jsq_over_Delta = (J_mean_0(omega_init))**2 / (Delta_hz)
    width_sym = r"\Omega_R" if pulse_shape == "square" else r"\sigma_f"

    fig, ax = plt.subplots(figsize=(8, 5.5))

    ax.plot(scan_freqs, pulse_pop, "o", color="C0", ms=4, label="simulation")
    if ok_line:
        xfine = np.linspace(scan_freqs[0], scan_freqs[-1], 2000)
        ax.plot(xfine, pulse_lineshape(xfine, *popt_line), "-", color="C0", lw=1.6,
                label=r"fit: $%s=%.2f$ Hz, $x_0=%.2f$ Hz"
                      % (width_sym, popt_line[1], f_center_rabi))
    ax.axvline(f_center_rabi, color="C3", ls="-", lw=1.1,
               label=r"fitted center $=%.2f$ Hz" % f_center_rabi)
    ax.axvline(f_drive_hz_init, color="C1", ls=":", lw=1.4,
               label=r"expected $=%.2f$ Hz" % f_drive_hz_init)
    ax.set_xlabel("drive frequency (Hz)")
    ax.set_ylabel(r"pop. in $%s$ after $%s$ pulse" % (ket(state_target), pulse_label))
    if pulse_shape == "gaussian":
        ax.set_title(r"profile, $\sigma_t=%.2f$ ms, total $=%.2f$ ms"
                     % (1000*gauss_sigma, pulse_duration_ms))
    else:
        ax.set_title(r"profile, $t_\mathrm{pulse}=%.2f$ ms" % pulse_duration_ms)
    ax.set_ylim(0, 1.05)
    ax.annotate(r"detuning $=%+.2f$ Hz" % detuning,
                xy=(0.02, 0.95), xycoords="axes fraction", va="top", ha="left",
                fontsize=10, bbox=dict(boxstyle="round", fc="w", ec="0.6", alpha=0.9))
    ax.legend(fontsize=8, loc="upper right")

    fig.suptitle(r"$%s \to %s$,  $V_0=%.2f$ Er,  $\alpha=[%s]$,  "
                 r"$\langle J\rangle^2/\Delta=%.1f$,  $n_\mathrm{avg}=%d$,  "
                 r"$\phi=%.2f$ rad,  pulse $=%s$"
                 % (ket(state_init), ket(state_target), V0, amp_str,
                    Jsq_over_Delta, n_floquet_avg, phi_floquet_avg, pulse_label),
                 fontsize=11)
    fig.tight_layout(rect=[0, 0, 1, 0.95])
    fig.savefig(fig_path, dpi=120)
    if show:
        plt.show()
    plt.close(fig)
    print("saved figure -> %s" % os.path.abspath(fig_path))

    return {
        "f_drive_hz_init": f_drive_hz_init,
        "f_center_rabi": f_center_rabi,
        "detuning": detuning,
        "pulse_duration_ms": pulse_duration_ms,
        "pulse_shape": pulse_shape,
        "gauss_trunc_sigma": gauss_trunc_sigma,
        "gauss_sigma_s": gauss_sigma,
        "gauss_area_s": gauss_area,
        "U_mean_0": U_mean_0,
        "U(V0)": U_V0,
        "N": N, "L": L, "sideband": sideband,
        "n_floquet_avg": n_floquet_avg,
        "phi_floquet_avg": phi_floquet_avg,
        "state_target": state_target,
        "mu_offset_hz": mu_site_hz.tolist(),
        "scan_freqs": scan_freqs, "pulse_pop": pulse_pop,
        "popt_lineshape": popt_line, "lineshape_ok": ok_line,
    }


# =============================================================
# Example call
# =============================================================
if __name__ == "__main__":
    res = resonance_Escan(
        state_init        = [1, 1],
        state_target      = [0, 2],
        V0_ER             = 7.5,
        Delta_hz          = 820.0,
        sideband          = -1,
        alpha             = [0.2],
        mu_offset_hz      = [],
        n_floquet_avg     = 1,
        phi_floquet_avg   = np.pi/2.0,
        pulse_duration_ms = 10.0,        # TOTAL pulse duration (ms), both shapes
        pulse_shape       = "square",    # 'square' or 'gaussian'
        gauss_trunc_sigma = 3.0,         # sigma = t_pulse / (2 * gauss_trunc_sigma)
        show              = True
    )
```
